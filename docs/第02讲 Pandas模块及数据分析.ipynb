### 2.2.1 创建 Series 对象

`pd.Series()` 是创建 Series 对象的核心函数。

| 参数 | 类型 | 说明 | 示例 |
|------|------|------|------|
| `data` | array-like, dict, scalar | 数据源：列表、ndarray、字典、标量等 | `[1, 2, 3]`、`np.array([1,2,3])`、`{'a':1,'b':2}`、`5` |
| `index` | array-like | 索引标签，默认 `RangeIndex(0,1,2...)` | `['a','b','c']`、`range(3)` |
| `dtype` | str / np.dtype | 数据类型，默认自动推断 | `'float64'`、`'int32'`、`np.float64` |
| `name` | str | Series 的名称 | `'温度'`、`'成绩'` |
| `copy` | bool | 是否复制输入数据，默认 `False` | `True` |

**常用创建方式一览：**

| 创建方式 | 说明 | 示例 |
|---------|------|------|
| `pd.Series([...])` | 从列表创建，索引自动生成 | `pd.Series([3.14, 1.3, 5.2])` |
| `pd.Series([...], index=[...])` | 指定自定义索引 | `pd.Series([1,2,3], index=['a','b','c'])` |
| `pd.Series(dict)` | 从字典创建，key→index，value→data | `pd.Series({'a':1, 'b':2})` |
| `pd.Series(scalar, index=[...])` | 用标量填充所有行 | `pd.Series(5, index=['a','b','c'])` |
| `pd.Series(ndarray)` | 从 NumPy 数组创建 | `pd.Series(np.random.randn(5))` |


#### 2.2.1.1 示例：多种方式创建 Series


````python
# 1. 从列表创建（默认整数索引）
s1 = pd.Series([3.14, 1.3, 5.2, -2.43])
s1
````

````python
# 2. 从列表创建，指定自定义索引、数据类型和名称
s2 = pd.Series(
    [3.14, 1.3, 5.2, -2.43],
    index=['第1行', '第2行', '第3行', '第4行'],
    dtype='float64',
    name='small series'
)
s2
````

````python
# 3. 自动生成'第i行'样式的索引
data = [3.14, 1.3, 5.2, -2.43, 5.33]
index = [f'第{i+1}行' for i in range(len(data))]
s3 = pd.Series(data, index=index, name='auto index')
s3
````

````python
# 4. 从字典创建（key 自动成为 index）
d = {'b': 1, 'a': 0, 'c': 2, 'e': 4}
s4 = pd.Series(d)
s4
````

````python
# 5. 从字典创建，并限制/重排 index
idx = list('abcde')
s5 = pd.Series(d, index=idx)  # 'z'不在idx中→NaN；缺值→NaN
s5
````

````python
# 6. 用标量填充所有行
s6 = pd.Series(5, index=['a', 'b', 'c', 'd', 'e'])
s6
````

````python
# 7. 从 NumPy 随机数组创建
s7 = pd.Series(np.random.randn(5), index=list('abcde'))
s7
````

#### 2.2.1.3 Series 与 NumPy 数组（ndarray）对比

| 对比维度 | Series | ndarray |
|---------|--------|--------|
| **索引** | 自带标签 index，可用字符串等任意类型 | 仅整数位置 0, 1, 2... |
| **切片** | 支持标签切片 `s['a':'c']` + 位置切片 | 仅位置切片 |
| **运算** | 按 index 自动对齐后再算 | 按位置逐元素计算 |
| **NumPy 兼容** | ✅ 可直接传入 NumPy 函数 | — |

> 📌 **一句话**：Series = ndarray + 标签索引。多出来的 index 让数据操作更直观、更安全。


````python
# Series 与 ndarray 比较：许多操作完全相同
s = pd.Series(np.random.randn(6), index=list('abcdef'))
print('Series s:'); print(s)

# 下标访问 —— 同 ndarray
print('\ns[0] =', s[0])

# 布尔掩码
print('\ns > 0:\n', s[s > 0])

# 花式索引
print('\ns[[0,2,4]]:\n', s[[0,2,4]])

# 通用函数（NumPy ufunc）
print('\nnp.exp(s):\n', np.exp(s))

# Series 特有的标签切片（左右均闭！）
print('\ns[\'a\':\'c\']:\n', s['a':'c'])
````

#### 2.2.1.4 Series 与 Python 字典对比

| 对比维度 | Series | dict |
|---------|--------|------|
| `'key' in obj` | ✅ 检查 index 中是否有该标签 | ✅ 检查 key 是否存在 |
| `obj['key']` | ✅ 通过 index 标签访问值 | ✅ 通过 key 访问值 |
| 切片 | ✅ 支持 | ❌ 不支持 |
| 向量化运算 | ✅ 支持 | ❌ 不支持 |
| 顺序 | ✅ 保持插入顺序 | ✅ Python 3.7+ 保持插入顺序 |


````python
# Series 可以像字典一样操作
s = pd.Series(np.random.randint(1,99,6), index=list('abcdef'))
print('s ='); print(s)

# in 操作 —— 检查 index
print('\n\'e\' in s:', 'e' in s)   # True
print('\'g\' in s:', 'g' in s)   # False

# 通过标签访问
print('\ns[\'b\']:', s['b'])
````

### 2.2.2 访问 Series 对象

Pandas 提供了多种灵活的方式访问 Series 中的元素：

| 访问方式 | 说明 | 示例 | 切片区间 |
|---------|------|------|---------|
| `s[idx]` | 直接下标/标签访问 | `s[0]`、`s['a']` | 下标左闭右开；标签左闭右闭 |
| `s.loc[idx]` | 基于**标签** index 访问 | `s.loc['a']`、`s.loc['a':'c']` | 左闭右闭 ✅ |
| `s.iloc[pos]` | 基于**整数位置**访问 | `s.iloc[0]`、`s.iloc[0:3]` | 左闭右开 |
| `s[mask]` | 布尔掩码访问 | `s[s > 50]` | — |
| `s[[idx1,idx2]]` | 花式索引（多标签/多位置） | `s[['a','c']]`、`s[[0,2]]` | — |

> ⚠️ **注意**：`s[]` 同时支持标签和位置，容易混淆。**推荐使用 `.loc[]`（标签）和 `.iloc[]`（位置）**，意图更明确。


#### 2.2.2.1 直接访问 `s[]`


````python
# 创建示例 Series
s = pd.Series(np.random.randint(1,99,6), index=['a','b','c','d','e','f'])
s
````

````python
# (1) 通过下标访问单个元素
print('s[1]:', s[1])       # 位置1的元素
print('s[3]:', s[3])       # 位置3的元素
````

````python
# (2) 通过标签访问单个元素
print("s['b']:", s['b'])
print("s['d']:", s['d'])
````

````python
# (3) 花式索引 —— 传入标签列表，返回子 Series
print("s[['b','d']]:"); print(s[['b','d']])
````

````python
# (4) 下标切片（左闭右开）
s[1:5]
````

````python
# (5) 标签切片（左右均闭！注意与列表切片不同）
s['b':'e']
````

````python
# (6) 布尔掩码 —— 选出 > 50 的元素
s[s > 50]
````

#### 2.2.2.2 `.loc[]` —— 基于标签 index 访问

`.loc[]` 严格按**索引标签**来定位数据，推荐用于标签明确的场景。


````python
# (1) 访问单个元素
s.loc['b']
````

````python
# (2) 标签切片（左右均闭）
s.loc['a':'c']
````

````python
# (3) 标签列表 —— 花式索引
s.loc[['a','c','e']]
````

````python
# (4) 布尔掩码
mask = [False, True, False, True, False, True]
s.loc[mask]
````

#### 2.2.2.3 `.iloc[]` —— 基于整数位置访问

`.iloc[]` 严格按**整数位置**（0, 1, 2...）来定位，行为与 NumPy 索引一致。


````python
# (1) 访问单个元素
s.iloc[1]
````

````python
# (2) 位置切片（左闭右开）
s.iloc[0:3]
````

````python
# (3) 位置列表 —— 花式索引
s.iloc[[0, 2, 4]]
````

````python
# (4) 布尔掩码
mask = [False, True, False, True, False, True]
s.iloc[mask]
````

### 2.2.3 修改 Series 对象

| 方法 | 参数 | 说明 | 示例 |
|------|------|------|------|
| `s[idx] = val` | — | 直接赋值修改单个元素 | `s['a'] = 100` |
| `s.replace()` | `to_replace, value, inplace` | 批量替换值（`inplace=False` 返回新对象） | `s.replace([1,2], [10,20])` |
| `s.rename()` | `index, inplace` | 重命名 index 标签 | `s.rename(index={'a':'AA'})` |
| `s.reindex()` | `labels` | 重排序/重索引（新标签无对应值→NaN） | `s.reindex(['c','b','a','x'])` |
| `s.name = '...'` | — | 修改 Series 名称 | `s.name = '成绩'` |
| `pd.concat([s1,s2])` | — | 拼接多个 Series（添加行） | `pd.concat([s1, s2])` |


#### 2.2.3.1 直接赋值修改值


````python
# 直接赋值 —— 通过标签或下标修改
s = pd.Series(np.random.randint(1,99,6), index=['a','b','c','d','e','f'])
print('原始:'); print(s)

# 通过标签修改
s['a'] = 100
print('\n修改 s[\'a\']=100 后:'); print(s)

# 通过下标修改
s[2] = 200
print('\n修改 s[2]=200 后:'); print(s)
````

#### 2.2.3.2 `replace()` 批量替换值


````python
# replace() —— 批量替换（inplace=False 返回新对象）
s1 = pd.Series([10, 30, 50, 70, 90], index=list('abcde'))
print('原始:'); print(s1)

# inplace=False（默认）：返回新对象，原对象不变
s2 = s1.replace(to_replace=[10,30,50], value=[100,300,500])
print('\nreplace 后 s1 (不变):'); print(s1)
print('\nreplace 后 s2 (新对象):'); print(s2)
````

````python
# inplace=True：原地修改，返回 None
s1.replace(to_replace=[10,30,50], value=[100,300,500], inplace=True)
print('inplace=True 后 s1:'); print(s1)
print('\n注意: inplace=True 返回 None，不要写成 s1 = s1.replace(...)')
````

#### 2.2.3.3 `rename()` 重命名索引


````python
# rename() —— 重命名 index 标签
s = pd.Series(np.random.randint(1,99,6), index=['a','b','c','d','e','f'])
print('原始:'); print(s)

# 通过字典映射重命名
s2 = s.rename(index={'a':'AA', 'c':'CC', 'e':'EE'})
print('\nrename 后:'); print(s2)
````

#### 2.2.3.4 `reindex()` 重排/重置索引


````python
# reindex() —— 按新顺序重排，不存在的标签填 NaN
s = pd.Series([10, 20, 30, 40, 50], index=list('abcde'))
print('原始:'); print(s)

# 重排索引顺序
s2 = s.reindex(['e','d','c','b','a'])
print('\n倒序 reindex:'); print(s2)

# 引入新标签（无对应值 → NaN）
s3 = s.reindex(['a','b','x','y','z'])
print('\n引入新标签:'); print(s3)
````

#### 2.2.3.5 修改 Series 名称


````python
# 修改 Series 的 name 属性
s = pd.Series([85, 92, 78, 88], index=['张三','李四','王五','赵六'])
print('原始 name:', s.name)

# 直接赋值修改
s.name = '期末成绩'
print('新  name:', s.name)
print(); print(s)
````

#### 2.2.3.6 添加元素 —— `pd.concat()` 拼接


````python
# 通过字典方式添加单个元素
s = pd.Series([10, 20, 30], index=['a','b','c'])
print('原始:'); print(s)

# 字典方式新增
s['d'] = 40
print('\n添加 d=40 后:'); print(s)
````

````python
# 通过 pd.concat() 拼接多个 Series
s1 = pd.Series([10, 20, 30], index=['a','b','c'])
s2 = pd.Series([100, 200], index=['x', 'y'])

# concat 合并
s3 = pd.concat([s1, s2])
print('s1:'); print(s1)
print('\ns2:'); print(s2)
print('\nconcat 后:'); print(s3)
````

### 2.2.4 删除 Series 对象

| 方法 | 参数 | 说明 | 示例 |
|------|------|------|------|
| `del s[idx]` | — | 删除单个元素（按标签），原地操作 | `del s['a']` |
| `s.drop(labels)` | `labels, axis, inplace` | 删除指定标签的元素（默认返回新对象） | `s.drop(['a','c'])` |
| `s[s != val]` | — | 布尔过滤「删除」（保留符合条件的行） | `s[s != 0]` |
| `s.dropna()` | — | 删除缺失值 NaN 所在元素 | `s.dropna()` |

> ⚠️ **注意**：`del` 只能按**标签**删除，不能按位置；`drop()` 默认 `inplace=False`，返回新对象。


#### 2.2.4.1 `del` —— 删除单个元素


````python
# del —— 按标签删除单个元素（原地操作）
s = pd.Series([10, 20, 30, 40, 50], index=['a','b','c','d','e'])
print('原始:'); print(s)

del s['c']  # 删除标签 'c' 对应的元素
print('\ndel s[\'c\'] 后:'); print(s)

# del s[1]  # ❌ 错误！del 不支持按位置删除
````

#### 2.2.4.2 `drop()` —— 删除一个或多个元素


````python
# drop() —— 按标签批量删除（默认返回新对象）
s = pd.Series([10, 20, 30, 40, 50], index=['a','b','c','d','e'])
print('原始:'); print(s)

# 删除多个标签
s2 = s.drop(['a', 'c'])
print('\ndrop([\'a\',\'c\']) 后:'); print(s2)
print('\n原 s 不变:'); print(s)

# inplace=True 原地删除
s.drop(['e'], inplace=True)
print('\ndrop inplace=True 后:'); print(s)
````

#### 2.2.4.3 布尔过滤「删除」


````python
# 布尔过滤 —— 保留满足条件的，等价于「删除」不满足的
s = pd.Series([10, 94, 30, 94, 50, 94], index=['a','b','c','d','e','f'])
print('原始:'); print(s)

# 「删除」值为 94 的元素（保留 != 94 的）
s_clean = s[s.values != 94]
print('\n过滤掉 94 后:'); print(s_clean)
````

### 2.3.1 创建 DataFrame 对象

`pd.DataFrame()` 是创建 DataFrame 对象的核心函数。

| 参数 | 类型 | 说明 | 示例 |
|------|------|------|------|
| `data` | ndarray, dict, list, DataFrame | 数据源 | `np.random.randn(8,4)`、`{'A':[1,2],'B':[3,4]}` |
| `index` | array-like | 行索引标签，默认 `RangeIndex` | `['a','b','c']`、`range(3)` |
| `columns` | array-like | 列标签，默认 `RangeIndex` | `['数学','语文','英语']` |
| `dtype` | np.dtype | 数据类型，默认自动推断 | `np.float64` |
| `copy` | bool | 是否复制数据，默认 `False` | `True` |

**常用创建方式一览：**

| 创建方式 | 说明 | 示例 |
|---------|------|------|
| `pd.DataFrame(ndarray)` | 从 NumPy 数组创建 | `pd.DataFrame(np.random.randn(8,4))` |
| `pd.DataFrame(dict)` | 从字典创建，key→列名 | `pd.DataFrame({'数学':[99,87],'语文':[85,91]})` |
| `pd.DataFrame(dict, index=[...])` | 指定行索引 | `pd.DataFrame(d, index=['张三','李四'])` |
| `pd.DataFrame(list_of_dicts)` | 从字典列表创建 | `pd.DataFrame([{'a':1,'b':2},{'a':3,'b':4}])` |
| `pd.DataFrame(Series_dict)` | 从 Series 字典创建 | `pd.DataFrame({'one':s1, 'two':s2})` |


#### 示例：多种方式创建 DataFrame


````python
# 1. 从 NumPy 数组创建（默认行列索引）
df1 = pd.DataFrame(np.random.randn(8, 4))
df1
````

### 2.4.2 CSV 文件操作

**什么是 CSV 文件？**

- CSV（Comma-Separated Values，逗号分隔值）是一种纯文本表格数据格式
- 每行一条记录，字段之间用逗号（或其他分隔符）隔开
- 通用性强：几乎所有数据处理软件（Excel、数据库、Python 等）都支持
- 文件扩展名通常为 `.csv`，也可用 `.txt`

**CSV 文件的打开方式：**

| 工具 | 说明 |
|------|------|
| 📝 文本编辑器 | 记事本、Notepad++、VS Code、Vim 等直接查看/编辑原始内容 |
| 📊 Excel / WPS | 以表格形式打开，直观浏览 |
| 🐍 Python Pandas | 用 `pd.read_csv()` 读入为 DataFrame，进行数据分析 |

**主要函数一览：**

| 函数 | 功能 | 说明 |
|------|------|------|
| `pd.read_csv()` | 📥 读取 CSV 文件 | 支持本地文件、URL、字符串 IO |
| `df.to_csv()` | 📤 写入 CSV 文件 | 将 DataFrame 保存为 CSV |
| `StringIO` | 🧵 内存字符串模拟文件 | 用于在内存中模拟 CSV 读写（无需实体文件） |


#### 2.4.2.1 写入 CSV 文件 —— `df.to_csv()`

**语法：**
```python
df.to_csv(path_or_buf=None, sep=',', na_rep='', columns=None,
          header=True, index=True, index_label=None, encoding=None)
```

**常用参数：**

| 参数 | 说明 | 示例 |
|------|------|------|
| `path_or_buf` | 文件路径或 StringIO 对象 | `'output.csv'` |
| `sep` | 分隔符，默认 `','` | `'	'`（制表符） |
| `encoding` | 编码格式，**中文推荐 `utf_8_sig`** | `'utf_8_sig'`、`'utf-8'` |
| `index` | 是否写入行索引，默认 `True` | `False` |
| `index_label` | 索引列的名称 | `'编号'` |
| `na_rep` | 缺失值填充字符 | `'NULL'` |

> ⚠️ 使用 `encoding='utf-8'` 时 Excel 直接打开可能乱码，推荐 `'utf_8_sig'`（带 BOM）。


````python
# 准备数据
names = ['宋江','卢俊义','吴用','公孙胜','关胜','林冲','秦明','呼延灼']
features = ['忠诚','智慧','勇敢','法术','生命']
heroes_df = pd.DataFrame(np.random.randint(90,100,(8,5)), index=names, columns=features)

# 写入 CSV —— 注意 encoding 用 utf_8_sig 避免 Excel 乱码
heroes_df.to_csv('data/heroes_table.csv', encoding='utf_8_sig')
print('CSV 已保存')
````

### 2.5.1 处理序列 Series

Series 缺失数据处理有三种核心方法：

| 方法 | 功能 | 说明 |
|------|------|------|
| `s.dropna()` | 🗑️ 删除 | 删除含 NaN 的元素，返回新 Series |
| `s.fillna(value)` | 📝 填充 | 用指定值或策略填充 NaN |
| `s.isna()` / `pd.isna(s)` | 🔍 检测 | 返回布尔掩码 Series，True 表示缺失 |


#### 2.5.1.1 删除缺失值 —— `s.dropna()`

| 参数 | 说明 | 示例 |
|------|------|------|
| `inplace` | 是否原地修改，默认 `False` | `inplace=True` |

> ⚠️ 空字符串 `''` **不是** NaN，`None` **是** NaN。


````python
# 创建含缺失值的 Series
s = pd.Series([1, np.nan, 5, np.nan, '', 4],
             index=['ele_'+str(i) for i in range(6)])
print('原始:'); print(s)

# dropna() —— 删除 NaN（注意 '' 不会被删除）
print('\ndropna 后:'); print(s.dropna())

# None 被视为 NaN
s['ele_2'] = None
print('\n设为 None 后 dropna:'); print(s.dropna())
````

#### 2.5.1.2 填充缺失值 —— `s.fillna()`

| 参数 | 说明 | 示例 |
|------|------|------|
| `value` | 填充的标量值 | `0`、`'missing'` |
| `method` | 填充策略：`'ffill'`(前向) / `'bfill'`(后向) | `method='ffill'` |
| `inplace` | 是否原地修改，默认 `False` | `inplace=True` |


````python
# fillna() —— 填充缺失值
s1 = s.copy()
print('原始:'); print(s1)

# 用 0 填充（inplace=False，返回新对象）
s2 = s1.fillna(0)
print('\nfillna(0) 后:'); print(s2)

# 原地填充（inplace=True，返回 None）
s1.fillna(0, inplace=True)
print('\n原地填充后 s1:'); print(s1)
````

#### 2.5.1.3 检测缺失值 —— `s.isna()` / `pd.isna()`

| 方法 | 说明 |
|------|------|
| `s.isna()` | 返回布尔 Series，NaN 对应 `True` |
| `pd.isna(s)` | 同上，函数形式 |
| `s.notna()` | 反向：非 NaN 对应 `True` |


````python
# isna() —— 检测缺失值
print('原始:'); print(s)
print('\nisna():'); print(s.isna())

# 利用掩码筛选非缺失值
print('\n非缺失元素:'); print(s[s.notna()])
````

### 2.5.3 实例讲解：处理缺失的风速数据

本案例使用某地多年风速观测数据，演示缺失数据处理的完整流程：

> 📥 读入 → 🔍 探查 → 🗑️ 方法一：删除 → 📐 方法二：利用列间关系插补 → 📊 前后对比


#### ① 读入数据并预处理


````python
# 读入风速数据（无表头，自定义列名）
windspeed = pd.read_excel('data/windspeed.xlsx', header=None,
    names=['year','month','day','Uday','Umean','Umeandir','Umax','Umaxdir'])

# 风速单位转换：原始数据是 0.1 m/s，除以 10 转为 m/s
for col in ['Uday', 'Umean', 'Umax']:
    windspeed[col] /= 10

# 整体概览
print('shape:', windspeed.shape)
print('\n统计摘要:\n', windspeed.describe())

# 前5年日风速趋势
windspeed['Uday'][:365*5].plot(figsize=(10,3), title='Daily Wind Speed (first 5 years)')
````

#### ② 方法一：直接删除 `dropna()` —— 效果不佳


````python
# 尝试直接删除含 NaN 的行
ws1 = windspeed.dropna(how='any')
print('原始行数:', len(windspeed))
print('删除后行数:', len(ws1))
print('\n删除后统计:\n', ws1.describe())

# 查看：1954 年的数据还在吗？
print('\n1954年数据:', len(ws1[ws1['year']==1954]), '行 ← 全部被删！')
````

> ⚠️ 发现 year 最小值变为 2003，1953-2002 的数据全部丢失！绝大多数行都有缺失值，直接删除不可行。


#### ③ 方法二：探索列间关系，利用比例插补

观察发现 **Uday（日均风速）列没有缺失**，可探索 Uday 与缺失列（Umean、Umax）的关系。


````python
# 先用不含 NaN 的子集观察相关性
ws_clean = windspeed.dropna()

import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 3, figsize=(14,4))
axes[0].plot(ws_clean['Uday'], ws_clean['Umean'], '.'); axes[0].set_title('Uday vs Umean')
axes[1].plot(ws_clean['Uday'], ws_clean['Umax'], '.'); axes[1].set_title('Uday vs Umax')
axes[2].plot(ws_clean['Umean'], ws_clean['Umax'], '.'); axes[2].set_title('Umean vs Umax')
plt.tight_layout(); plt.show()

print('相关系数:')
print(ws_clean[['Uday','Umean','Umax']].corr())
````

#### ④ 计算比例系数并进行插补


````python
# 计算比例系数（处理除零 → inf → NaN）
ratio_mean = (windspeed['Umean'] / windspeed['Uday']).replace([np.inf, -np.inf], np.nan).mean()
ratio_max  = (windspeed['Umax']  / windspeed['Uday']).replace([np.inf, -np.inf], np.nan).mean()
print(f'Umean/Uday = {ratio_mean:.3f}')
print(f'Umax/Uday  = {ratio_max:.3f}')

# 用比例关系填充缺失值
ws_filled = windspeed.copy()
ws_filled.loc[ws_filled['Umean'].isna(), 'Umean'] = \
    ws_filled.loc[ws_filled['Umean'].isna(), 'Uday'] * ratio_mean
ws_filled.loc[ws_filled['Umax'].isna(), 'Umax'] = \
    ws_filled.loc[ws_filled['Umax'].isna(), 'Uday'] * ratio_max

print('\n填充后统计:\n', ws_filled.describe())
````

#### ⑤ 填充前后对比 —— 以 1954 年为例


````python
# 1954 年：填充前 vs 填充后
before = windspeed[windspeed['year']==1954]
after  = ws_filled[ws_filled['year']==1954]

fig, axes = plt.subplots(1, 2, figsize=(12,4))
axes[0].plot(before[['Uday','Umean']]); axes[0].set_title('1954 Before (Umean missing)')
axes[0].legend(['Uday','Umean'])
axes[1].plot(after[['Uday','Umean']]); axes[1].set_title('1954 After (Umean filled)')
axes[1].legend(['Uday','Umean'])
plt.tight_layout(); plt.show()
````

### 2.5.4 思考：如何处理噪声数据？

除了缺失值，真实数据中常包含**噪声（异常值）**。本小节介绍几种常用处理方法。


#### ① 读入数据并观察


````python
import matplotlib.pyplot as plt

# 读入时间序列数据
timedata = pd.read_excel('data/时间序列.xlsx',
    index_col=None, header=None, names=['time','data'])

# 全局视图 + 放大视图
fig, axes = plt.subplots(1, 2, figsize=(12,4))
timedata['data'].plot(ax=axes[0]); axes[0].set_title('Full View')
axes[1].plot(timedata['time'], timedata['data'])
axes[1].set_ylim([-3, 3]); axes[1].set_title('Zoomed In')
plt.tight_layout(); plt.show()
````

#### ② 寻找异常点


````python
# 方法1：统计摘要 + 直方图
print(timedata['data'].describe())
timedata['data'].hist(bins=30, figsize=(8,3))
plt.title('Data Distribution'); plt.show()
````

````python
# 方法2：箱形图（基于四分位数识别异常值）
import seaborn as sns
sns.boxplot(x=timedata['data'], color='lime')
plt.title('Box Plot'); plt.show()
````

````python
# 方法3：3σ 原则（超过均值±3倍标准差视为异常）
mu, sigma = timedata['data'].mean(), timedata['data'].std()
lower, upper = mu - 3*sigma, mu + 3*sigma
condition = (timedata['data'] < lower) | (timedata['data'] > upper)
print(f'均值={mu:.3f}, 标准差={sigma:.3f}')
print(f'异常范围: <{lower:.3f} 或 >{upper:.3f}')
print(f'异常点数: {condition.sum()} / {len(timedata)}')
````

#### ③ 噪声处理方法对比

| 方法 | 说明 | 适用场景 |
|------|------|------|
| 均值填充 | 用均值替换异常值 | 异常点少、数据平稳 |
| 前后填充 | 用前/后一个正常值填充 | 时序数据，相邻值接近 |
| 低通滤波 | 用信号处理滤除高频噪声 | 噪点密集、需平滑整体趋势 |


````python
# --- 方法A：均值填充 ---
td1 = timedata.copy()
td1.loc[condition, 'data'] = td1['data'].mean()

# --- 方法B：前后填充（先置 NaN 再 ffill） ---
td2 = timedata.copy()
td2.loc[condition, 'data'] = np.nan
td2['data'] = td2['data'].fillna(method='pad')

# 对比图
fig, axes = plt.subplots(1, 3, figsize=(14,3))
timedata['data'].plot(ax=axes[0]); axes[0].set_title('Original')
td1['data'].plot(ax=axes[1]); axes[1].set_title('Mean Fill')
td2['data'].plot(ax=axes[2]); axes[2].set_title('Forward Fill')
for ax in axes: ax.set_ylim([-3,3])
plt.tight_layout(); plt.show()
````

````python
# --- 方法C：低通滤波（Butterworth） ---
from scipy import signal
b, a = signal.butter(8, 0.01, 'lowpass')
td3 = timedata.copy()
td3['filtered'] = signal.filtfilt(b, a, td3['data'])

plt.figure(figsize=(10,3))
plt.plot(timedata['time'], timedata['data'], alpha=0.4, label='Original')
plt.plot(td3['time'], td3['filtered'], 'r-', label='Filtered')
plt.legend(); plt.ylim([-2,2]); plt.title('Low-pass Filter'); plt.show()
````

### 2.6.6 案例 —— 风速数据分析

综合运用统计方法分析风速数据，包括风向玫瑰图和极值预测。


#### 风速风向玫瑰图


````python
# 读入风速数据
windspeed = pd.read_excel('data/windspeed.xlsx', header=None,
    names=['year','month','day','Uday','Umean','Umeandir','Umax','Umaxdir'])

# 计算各风向的最大风速
mean_U = [windspeed[windspeed['Umeandir']==i]['Umean'].max()/10 for i in range(1,17)]
max_U  = [windspeed[windspeed['Umaxdir']==i]['Umax'].max()/10   for i in range(1,17)]
mean_U.append(mean_U[0])  # 闭合极坐标
max_U.append(max_U[0])

angle = np.linspace(0, 2*np.pi, 17)

# 极坐标绘图
fig, ax = plt.subplots(1, 1, subplot_kw={'projection':'polar'})
ax.plot(angle, mean_U, 'b*-', label='Mean Wind')
ax.plot(angle, max_U,  'r*-', label='Max Wind')
ax.set_theta_zero_location('N'); ax.set_theta_direction(-1)
ax.legend(); plt.title('Wind Rose'); plt.show()
````

#### 100 年一遇极值风速预测（Gumbel 分布）

利用年最大风速拟合 Gumbel 极值分布，预测 100 年重现期的极值风速。


````python
from scipy.stats import gumbel_r

# 年最大风速
annual_max = windspeed.groupby('year')['Umean'].max().dropna() / 10

# 拟合 Gumbel 分布
params = gumbel_r.fit(annual_max)

# 预测 100 年一遇风速
p_100yr = 1 - 1/(100*365)
ws_100yr = gumbel_r.ppf(p_100yr, *params)
print(f'100年一遇风速: {ws_100yr:.2f} m/s')
print(f'历史最大风速:   {annual_max.max():.2f} m/s')

# 拟合效果图
fig, ax = plt.subplots(1,1)
annual_max.hist(bins=20, density=True, alpha=0.5, ax=ax)
x = np.linspace(annual_max.min(), annual_max.max()*1.2, 100)
ax.plot(x, gumbel_r.pdf(x, *params), 'r-', lw=2, label='Gumbel Fit')
ax.legend(); ax.set_title('Annual Max Wind Speed'); plt.show()
````

| 参数 | 说明 | 示例 |
