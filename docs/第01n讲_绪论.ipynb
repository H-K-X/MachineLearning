## 1.2 什么是机器学习？

### 1.2.1 机器学习的定义

> **机器学习（Machine Learning）** 是人工智能的核心分支，研究如何让计算机系统**从数据中自动学习规律与模式**，并利用所学知识对未知数据进行预测或决策，而无需人为显式地编程每一条规则。
>
> 这一概念最早由 **Arthur Samuel**（1959）提出：
>
> *"Machine Learning is the field of study that gives computers the ability to learn without being explicitly programmed."*
>
> —— 机器学习赋予计算机**无需显式编程即可学习**的能力。
>
> 目前最广泛引用的学术定义来自 **Tom Mitchell**（1997）：
>
> *"A computer program is said to **learn** from experience **E** with respect to some class of tasks **T** and performance measure **P**, if its performance at tasks in **T**, as measured by **P**, improves with experience **E**."*
>
> —— 对于某类**任务 T** 和**性能度量 P**，若计算机程序在 T 上的表现（以 P 衡量）随**经验 E** 的增加而提升，则称该程序从经验 E 中**学习**。
>
> **📌 通俗公式：机器学习 = 数据 + 算法 → 模型 → 预测/决策**

### 1.2.2 AI — 机器学习 — 深度学习

> **三者关系：AI ⊃ 机器学习 ⊃ 深度学习**（由外到内的包含关系）

- **人工智能 (Artificial Intelligence）**  
  利用计算机解决人类通过直觉可以解决的问题（如：自然语言理解、图像识别、语音识别等）—— **最外层的宏大目标**

- **机器学习 (Machine Learning)**    
  通过赋予机器学习的能力使其完成一系列任务，并用于决策，但决策可靠性可能需要人类干预 —— **实现AI的核心方法**

- **深度学习 (Deep Learning)**    
  基于复杂人工神经网络，自动将简单特征组合成更复杂的特征，并用这些特征解决问题，且无需人类干预 —— **当前最强有力的ML技术**

> 📌 简言之：**深度学习 ⊂ 机器学习 ⊂ 人工智能**，深度学习是目前推动AI浪潮的核心引擎。

### 1.2.3 机器学习 (ML) 是AI的核心

- **为什么ML是当代AI的核心？** 传统AI依赖人工编写规则（专家系统），难以应对复杂、多变的真实场景。机器学习让计算机**从数据中自动发现规律**，使AI具备了可扩展性和泛化能力。
- **数据驱动**：大数据时代的到来为ML提供了海量"燃料"，模型的性能随数据量增加而持续提升。
- **三大要素**：**数据 + 算法 + 算力**共同推动了本轮AI浪潮，而机器学习正是连接三者的桥梁。
- **实际体现**：从语音识别、图像分类到ChatGPT等大语言模型，其背后无一不是机器学习（尤其是深度学习）在驱动。

- 机器学习领域的Hello World!


```python
# 【Hello World】机器学习入门示例 —— 鸢尾花分类
# 仅需几行代码，体验机器学习的完整流程
import matplotlib.pyplot as plt
import matplotlib as mpl
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
#plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'DejaVu Sans']
#plt.rcParams['axes.unicode_minus'] = False

# 设置中文字体
mpl.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'DejaVu Sans']
mpl.rcParams['axes.unicode_minus'] = False

# ============================================================
# 1. 加载经典数据集
# ============================================================
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=42
)

# ============================================================
# 2. 训练 KNN 分类模型
# ============================================================
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# ============================================================
# 3. 预测并评估
# ============================================================
y_pred = model.predict(X_test)
accuracy = model.score(X_test, y_test)
print(f"✅ 鸢尾花分类准确率: {accuracy:.2%}")
print(f"💡 仅用 {len(X_train)} 条数据训练，模型就能以 {accuracy:.2%} 的准确率识别3种鸢尾花！")
print(f"📌 这就是机器学习的魅力——让计算机从数据中自动学习规律。")

# ============================================================
# 4. 可视化 —— 一图胜千言
# ============================================================
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# 选取前两个特征（花萼长度、花萼宽度）用于可视化
feature_names = iris.feature_names[:2]
X_all = iris.data[:, :2]
y_all = iris.target
target_names = iris.target_names
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

# --- 子图1：全量数据分布 ---
for i, (name, color) in enumerate(zip(target_names, colors)):
    mask = y_all == i
    axes[0].scatter(X_all[mask, 0], X_all[mask, 1],
                    c=color, label=name, edgecolors='white', s=50, alpha=0.8)
axes[0].set_xlabel(feature_names[0], fontsize=11)
axes[0].set_ylabel(feature_names[1], fontsize=11)
axes[0].set_title('鸢尾花数据集 (150条)', fontsize=13, fontweight='bold')
axes[0].legend(loc='upper right')

# --- 子图2：训练集 + 分类决策边界 ---
axes[1].scatter(X_train[:, 0], X_train[:, 1], c=[colors[t] for t in y_train],
                edgecolors='white', s=50, alpha=0.8)
axes[1].set_xlabel(feature_names[0], fontsize=11)
axes[1].set_title('训练集 (105条) — 模型从中学习', fontsize=13, fontweight='bold')

# --- 子图3：测试集预测结果 ---
correct = y_pred == y_test
for i, (name, color) in enumerate(zip(target_names, colors)):
    # 预测正确的点
    mask_correct = (y_test == i) & correct
    axes[2].scatter(X_test[mask_correct, 0], X_test[mask_correct, 1],
                    c=color, marker='o', s=80, edgecolors='white',
                    label=f'{name} ✓')
    # 预测错误的点
    mask_wrong = (y_test == i) & ~correct
    if mask_wrong.any():
        axes[2].scatter(X_test[mask_wrong, 0], X_test[mask_wrong, 1],
                        c=color, marker='X', s=120, edgecolors='black',
                        linewidths=1.5, label=f'{name} ✗')
axes[2].set_xlabel(feature_names[0], fontsize=11)
axes[2].set_title(f'测试集预测结果 (准确率 {accuracy:.1%})', fontsize=13, fontweight='bold')
axes[2].legend(loc='upper right')

plt.tight_layout()
plt.show()
```

#### （1）唐纳德·赫布（Donald Hebb, 1904–1985）—— 加拿大著名生理心理学家

- **Hebb 学习规则**（1949）：基于神经心理学的学习机制，开启了机器学习的第一步，此后被称为 **Hebb 学习规则**。
- **核心思想**：Hebb 是一种**无监督学习**规则，其理论可概括为——**"一起激发的神经元，连接在一起"**（*Cells that fire together, wire together.*）

**经典实验 —— 巴甫洛夫的狗**

 <img src='images/ch01/traindog.png' width=500>

 铃声响时一个神经元被激发，同时食物出现激发附近的另一个神经元，两个神经元间的联系就会**强化**，从而记住两件事之间的关联。反之，若总是不同步激发，联系就会**弱化**。

#### （2）阿兰·图灵（Alan Turing, 1912–1954）—— 计算机科学之父、人工智能先驱

- 1950 年，图灵发表了划时代论文《Computing Machinery and Intelligence》，提出了著名的 **"图灵测试"（Turing Test）** 来判定机器是否具有智能。
- **图灵测试**的核心思想：如果一台机器能够与人类展开对话（通过电传设备），而人类无法辨别对方是机器还是人，则称这台机器具有**智能**。
- 这一思想实验令人信服地论证了"**思考的机器**"是可能的，为人工智能的诞生奠定了哲学基础。

> 📌 图灵测试至今仍是衡量 AI 智能水平的经典参照标准。

<img src='images/ch01/tuling.png' width=220>

#### （3）赫伯特·西蒙（Herbert A. Simon, 1916–2001）—— 诺贝尔经济学奖得主、人工智能奠基人之一

- 赫伯特·西蒙是罕见的跨学科大师，身兼**计算机科学家、经济学家、心理学家**，1975 年获图灵奖，1978 年获诺贝尔经济学奖。
- 他对**学习**给出了一个至今仍被引用的精确定义：

> *"学习表示系统中的自适应变化，该变化能使系统比上一次更有效地完成同一群体所执行的同样任务。"*

- 这一定义强调了学习的三个关键要素：**自适应变化**、**效率提升**、**可重复任务**。

#### （4）阿瑟·塞缪尔（Arthur Samuel, 1901–1990）—— "机器学习之父"

- 1952 年，IBM 的 Arthur Samuel 在论文《Some Studies in Machine Learning Using the Game of Checkers》中首次定义并使用了 **"机器学习"（Machine Learning）** 一词，被后人誉为 **"机器学习之父"**。

> 📜 **Samuel 对机器学习的经典定义：**
>
> *"在不直接针对问题进行编程的情况下，赋予计算机学习能力的一个研究领域。"*

- 他还编写了一款**会下西洋跳棋的程序**——通过观察棋子的走位构建模型，不断提升棋艺。Samuel 与自己对弈多场后发现，程序越下越好。
- 这是历史上第一个**通过自我对弈实现能力进化**的 AI 程序，开创了"从经验中学习"的实践先河。

#### （5）弗兰克·罗森布拉特（Frank Rosenblatt, 1928–1971）—— 感知机之父

- 1957 年，Rosenblatt 发明了 **感知机（Perceptron）**，这是**第一个真正意义上的人工神经网络**，也是后来支持向量机（SVM）的思想源头，在当时引起了巨大轰动。
- **感知机**是一种二分类的线性分类模型：
  - **输入**：实例的特征向量
  - **输出**：+1 或 −1（正类/负类）
  - **本质**：在特征空间中寻找一个分离超平面，将正负两类样本线性划分
- 感知机的诞生标志着**连接主义（Connectionism）** 学派的正式确立。

<img src='images/ch01/perceptron.png' width=280>

#### （6）马文·明斯基（Marvin Minsky, 1927–2016）—— "人工智能之父"

- 麻省理工学院（MIT）人工智能实验室联合创始人，美国工程院与科学院双料院士，人工智能领域先驱。
- 1969 年，Minsky 与 Papert 合著《Perceptrons》，在书中提出了著名的 **XOR 问题**，严格证明了单层感知机**无法处理线性不可分数据**。
- 这一论证指出了感知机的根本局限，直接导致神经网络研究进入长达十余年的 **"AI 寒冬"（AI Winter）**——研究经费骤减、学界热情冷却。
- Rosenblatt 在此后两年不幸离世，令人唏嘘。直到 1980 年代多层网络和反向传播算法兴起，连接主义才重获新生。

<img src='images/ch01/05.png' width=300>

#### （7）杰弗里·辛顿（Geoffrey Hinton, 1947– ）—— "深度学习之父"

- 2006 年，Hinton 与其学生 Salakhutdinov 在 **《Nature》** 上发表里程碑论文，提出 **深度置信网络（Deep Belief Network, DBN）**，正式开启了 **深度学习新纪元**。
- 2012 年，Hinton 团队以 AlexNet 在 ImageNet 图像识别大赛中以**压倒性优势夺冠**，深度学习从此成为 AI 领域的主流范式。
- 2018 年，Hinton 与 Yann LeCun、Yoshua Bengio 共同获得 **图灵奖**，被誉为 **"深度学习三巨头"**。

- **Yann LeCun**（中文名：杨立昆），Facebook 人工智能实验室（FAIR）主任，纽约大学终身教授，2018 年图灵奖得主。
- 早在上世纪 80 年代就开始研究神经网络，是**将多层神经网络推进到深度学习概念的第一人**。
- 核心贡献：发明了 **卷积神经网络（CNN）**，在手写数字识别（MNIST）、图像分类等领域取得了突破性成果，CNN 至今仍是计算机视觉的基石架构。

> 🏆 2018 年图灵奖授予 **Hinton、LeCun、Bengio** 三位深度学习先驱，表彰他们为 AI 做出的革命性贡献。

## 1.4 机器学习的分类
（按照是否在人的监督下进行训练分类）

- 监督学习

    - 任务流程：给定数据集，训练预测函数，对于新未知数据，预测出结果
    - 数据要求：包括输入（亦称特征）和输出（亦称目标）
    - 目标数据：目标是由人标注
    - 问题分类：统计分类、回归分析
    - 数  据  集：训练集 + 测试集


- 无监督学习

    - 特别之处：不同于监督学习，无监督学习没有人为标注的目标数据
    - 常见算法：聚类、生成对抗网络（GAN）等
    

- 半监督学习 —— 介于监督学习与无监督学习之间


- 增强学习
    - 行为调整：为了达成目标，随着环境的变动，逐步调整机器学习
    - 评估反馈：每一个行为之后的指标（正向或负向）

## 1.5 机器学习算法概览

> 本课程侧重**应用与实践**，不深入讲解算法理论。下面给出常见算法的分类总览，建立整体认知即可。

| 类别 | 代表算法 | 典型应用 |
|------|---------|---------|
| **分类** | KNN、决策树、SVM、朴素贝叶斯 | 垃圾邮件识别、图像分类 |
| **回归** | 线性回归、岭回归、Lasso | 房价预测、趋势分析 |
| **聚类** | K-Means、DBSCAN、层次聚类 | 客户分群、图像分割 |
| **降维** | PCA、t-SNE | 数据可视化、特征压缩 |
| **集成学习** | 随机森林、AdaBoost、XGBoost | 竞赛刷榜、工业落地首选 |
| **神经网络/深度学习** | MLP、CNN、RNN、Transformer | 计算机视觉、NLP、大模型 |

> 📌 我们将在后续课程中逐一实践上述算法（使用 Scikit-Learn 等工具包）。
